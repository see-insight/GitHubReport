```python
extra_repos = [
    # "student1/private‑capstone",
    # "student2/side‑project",
    # "another‑org/shared‑repo",
]
```

```python
import GitHubReport as GHR


provider = GHR.GitHubProvider(
    org=ORG,
    token=token1,
    extra_repos=extra_repos,
)
```

```python
group = GHR.Group(
    provider=provider,
    start=SEMESTER_START,
    end=SEMESTER_END,
)
```

```python
summary = group.student(USERNAME)
```

```python
group.plot_student_activity_calendar(summary)
```
